```python
import pandas as pd
from sklearn.linear_model import LinearRegression

# Create dataset
data = {
    'Area': [1000, 1500, 2000, 2500, 3000],
    'Price': [2000000, 3000000, 4000000, 5000000, 6000000]
}

df = pd.DataFrame(data)

# Input and output
X = df[['Area']]
y = df['Price']

# Train model
model = LinearRegression()
model.fit(X, y)

# Predict house price
new_house = pd.DataFrame({'Area': [1800]})
prediction = model.predict(new_house)

print("House Area:", new_house['Area'][0], "sq.ft")
print("Predicted House Price: Rs.", prediction[0])
```
